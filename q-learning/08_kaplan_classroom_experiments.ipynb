```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# ========== Section 1: Data Loading and Plotting ==========

def load_and_clean_data(filepath):
    print("Loading and cleaning data...")
    data = pd.read_csv(filepath)
    cleaned_data = data[(data['matched_info'] == 'yes') & 
                        (data['bid_fp'] <= data['value_fp']) & 
                        (data['bid_fp'] >= 0) &
                        (data['value_fp'] <= 30) &
                        (data['value_fp'] >= 0)]
    print(f"Loaded and cleaned {len(cleaned_data)} rows.")
    return cleaned_data

def plot_actual_bids(filepath, axes):
    data = load_and_clean_data(filepath)
    
    # Regression: bid on valuation and valuation^2 without a constant
    data['valuation_squared'] = data['value_fp'] ** 2
    X = data[['value_fp', 'valuation_squared']]  # Independent variables
    y = data['bid_fp']  # Dependent variable (bids)
    
    model = sm.OLS(y, X).fit()  # Regression without a constant
    
    # Valuation grid for regression line
    valuation_grid = np.linspace(0, 30, 100)
    predicted_bids = model.predict(np.column_stack((valuation_grid, valuation_grid ** 2)))
    
    # Plot actual bids and the regression line on top-left
    axes[0, 0].scatter(data['value_fp'], data['bid_fp'], color='blue', alpha=0.5, label='Actual Bids')
    axes[0, 0].plot(valuation_grid, predicted_bids, color='red', label='Regression Line')
    axes[0, 0].set_xlabel('Valuation')
    axes[0, 0].set_ylabel('Bid')
    axes[0, 0].set_title('Actual Bids vs Valuation')
    axes[0, 0].legend()
    axes[0, 0].set_ylim(0, 30)

    # Top-right: Distribution of Bids (KDE)
    sns.kdeplot(data['bid_fp'], ax=axes[0, 1], fill=True, color='blue')
    axes[0, 1].set_xlabel('Bid')
    axes[0, 1].set_ylabel('Density')
    axes[0, 1].set_title('Distribution of Bids')
    axes[0, 1].set_xlim(0, 30)
    
    return model, valuation_grid, predicted_bids

# ========== Section 2: Q-Learning Agents and Auction Setup ==========

class QLearningAgent:
    def __init__(self, valuation_distribution=lambda: np.random.uniform(0, 1), action_min=0, action_max=1, granularity=20, epsilon=0.1, alpha=0.1, min_val=0, max_val=1, epsilon_decay=None):
        self.valuation_distribution = valuation_distribution
        self.actions = np.linspace(action_min, action_max, granularity)
        self.epsilon = epsilon
        self.alpha = alpha
        self.min_val = min_val
        self.max_val = max_val
        self.granularity = granularity
        self.q_values = np.zeros((granularity, len(self.actions)))
        self.valuation = self.valuation_distribution()
        self.state = self.get_state(self.valuation)

    def get_state(self, valuation):
        bin_centers = np.linspace(self.min_val, self.max_val, self.granularity)
        return np.argmin(np.abs(bin_centers - valuation))
    
    def refresh_valuation(self):
        self.valuation = self.valuation_distribution()
        self.state = self.get_state(self.valuation)
    
    def choose_action(self):
        if np.random.rand() < self.epsilon:
            return np.random.choice(self.actions)
        else:
            return self.actions[np.argmax(self.q_values[self.state])]

    def update_q_values(self, chosen_action, reward):
        action_index = np.where(self.actions == chosen_action)[0][0]
        self.q_values[self.state, action_index] += self.alpha * (reward - self.q_values[self.state, action_index])

    def feedback(self, reward, action):
        self.update_q_values(action, reward)

class SingleSidedAuction:
    def __init__(self, agents, price_floor=0, price_ceiling=1, payment_rule=None, b=1.0, c=0.0):
        self.agents = agents
        self.price_floor = price_floor
        self.price_ceiling = price_ceiling
        self.payment_rule = payment_rule
        self.b = b
        self.c = c

    def run_auction(self):
        for agent in self.agents: agent.refresh_valuation()
        bids = [max(self.price_floor, min(agent.choose_action(), self.price_ceiling)) for agent in self.agents]
        winner_index, payment = self.payment_rule(bids)
        
        for idx, agent in enumerate(self.agents):
            if idx == winner_index:
                payoff = agent.valuation - payment
                reward = self.b * payoff + self.c * payoff ** 2 if payoff > 0 else payoff
                agent.feedback(reward, bids[idx])
            else:
                agent.feedback(0, bids[idx])  # Non-winners get a reward of 0

def run_single_simulation(agents, game, num_rounds=10000):
    for _ in range(num_rounds): game.run_auction()
    return [agent.q_values for agent in agents]

def run_monte_carlo_simulation(agents, auction, num_simulations=30, num_rounds=10000):
    all_bids_for_valuations = [[] for _ in range(len(agents))]
    for sim in range(num_simulations):
        print(f"Running simulation {sim+1}/{num_simulations}...")
        run_single_simulation(agents, auction, num_rounds=num_rounds)
        for i, agent in enumerate(agents):
            avg_bids_for_valuation = [agent.actions[np.argmax(agent.q_values[state])] for state in range(agent.granularity)]
            all_bids_for_valuations[i].append(avg_bids_for_valuation)

    percentiles = {}
    for i in range(len(agents)):
        avg_bids_for_agent = np.array(all_bids_for_valuations[i])
        percentiles[i] = {
            "5th": np.percentile(avg_bids_for_agent, 5, axis=0),
            "median": np.median(avg_bids_for_agent, axis=0),
            "95th": np.percentile(avg_bids_for_agent, 95, axis=0),
        }
    print("Monte Carlo simulations completed.")
    return percentiles

# ========== Section 3: Plotting Simulated Results ==========

def plot_simulation_overlay(filepath, percentiles, v_guess, axes, valuation_grid, predicted_bids, model):
    # Load true bids and valuations
    true_valuations, true_bids = get_true_bids(filepath)

    # Scale simulated percentiles to the range [0, 30]
    valuations_scaled = np.linspace(0, 30, len(percentiles[0]["median"]))
    simulated_bids_scaled = 30 * percentiles[0]["median"]

    # Bottom-left: Plot observed vs simulated bidding functions with actual data points
    axes[1, 0].plot(valuation_grid, predicted_bids, label='Regression Line', color='red')
    axes[1, 0].scatter(true_valuations, true_bids, label='Actual Bids', color='green', s=10, alpha=0.6)
    axes[1, 0].plot(valuations_scaled, simulated_bids_scaled, label='Simulated Median Bidding Function (Scaled)', color='blue', linestyle='--')
    axes[1, 0].set_xlabel('Valuation')
    axes[1, 0].set_ylabel('Bid')
    axes[1, 0].set_title('Observed vs Simulated Bidding Functions with Actual Bids')
    axes[1, 0].legend()
    axes[1, 0].set_ylim(0, 30)

def plot_kde_comparison(filepath, percentiles, axes):
    # Load true bids and valuations
    true_valuations, true_bids = get_true_bids(filepath)

    # Scale simulated percentiles to the range [0, 30]
    simulated_bids_scaled = 30 * percentiles[0]["median"]

    # Bottom-right: Plot KDE for observed and simulated bids
    sns.kdeplot(true_bids, label='Observed Bids', color='red', fill=True, ax=axes[1, 1])
    sns.kdeplot(simulated_bids_scaled, label='Simulated Bids', color='blue', linestyle='--', fill=True, ax=axes[1, 1])
    axes[1, 1].set_xlabel('Bid')
    axes[1, 1].set_ylabel('Density')
    axes[1, 1].set_title('KDE of Observed vs Simulated Bids')
    axes[1, 1].legend()

# ========== Example Usage ==========

v_values = [30]  # Guessed valuation ranges
filepath = '/home/pp712/double-auctions/code/14_simple/clean_kaplan_data.csv'

# Set up the figure and axes for 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# First: Load and plot the actual bids and regression line
print("Stage 1: Load and plot actual bids.")
model, valuation_grid, predicted_bids = plot_actual_bids(filepath, axes)

# Test different reward function parameter c values
c_values = [-0.5, 0.0, 0.5]

for c in c_values:
    print(f"\nTesting for c={c}...")

    # Second: Set up Q-learning agents
    print("Stage 2: Set up Q-learning agents and run simulations.")
    num_agents = 2
    agents = [QLearningAgent() for _ in range(num_agents)]

    # Set up auction with custom reward function
    auction = SingleSidedAuction(agents, payment_rule=first_price_rule, b=1.0, c=c)

    # Run Monte Carlo simulations
    percentiles = run_monte_carlo_simulation(agents, auction, num_simulations=30, num_rounds=20000)

    # Plot simulation results
    print("Stage 3: Overlay simulated results on actual data.")
    plot_simulation_overlay(filepath, percentiles, v_guess=30, axes=axes, valuation_grid=valuation_grid, predicted_bids=predicted_bids, model=model)

    # Plot KDE comparison between observed and simulated bids
    print("Stage 4: Plot KDE comparison of observed and simulated bids.")
    plot_kde_comparison(filepath, percentiles, axes=axes)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

# ========== Section 1: Data Loading and Plotting for All-Pay Auction ==========

def load_and_clean_all_pay_data(filepath):
    data = pd.read_csv(filepath)
    cleaned_data = data[(data['matched_info'] == 'yes') & 
                        (data['bid_ap'] >= 0) &
                        (data['bid_ap'] <= 30) &
                        (data['value_ap'] <= 30) &
                        (data['value_ap'] >= 0)]
    return cleaned_data

def plot_all_pay_bids(filepath, axes):
    data = load_and_clean_all_pay_data(filepath)
    
    # Regression: bid on valuation and valuation^2 without a constant
    data['valuation_squared'] = data['value_ap'] ** 2
    X = data[['value_ap', 'valuation_squared']]  # Independent variables
    y = data['bid_ap']  # Dependent variable (bids)
    
    model = sm.OLS(y, X).fit()  # Regression without a constant
    
    # Valuation grid for regression line
    valuation_grid = np.linspace(0, 30, 100)
    predicted_bids = model.predict(np.column_stack((valuation_grid, valuation_grid ** 2)))
    
    # Plot actual bids and the regression line on top-left
    axes[0, 0].scatter(data['value_ap'], data['bid_ap'], color='blue', alpha=0.5, label='Actual Bids')
    axes[0, 0].plot(valuation_grid, predicted_bids, color='red', label='Regression Line')
    axes[0, 0].set_xlabel('Valuation')
    axes[0, 0].set_ylabel('Bid')
    axes[0, 0].set_title('All-Pay Bids vs Valuation')
    axes[0, 0].legend()
    axes[0, 0].set_ylim(0, 30)

    # Top-right: Distribution of Bids (KDE)
    sns.kdeplot(data['bid_ap'], ax=axes[0, 1], fill=True, color='blue')
    axes[0, 1].set_xlabel('Bid')
    axes[0, 1].set_ylabel('Density')
    axes[0, 1].set_title('Distribution of All-Pay Bids')
    axes[0, 1].set_xlim(0, 30)
    
    return model, valuation_grid, predicted_bids

# ========== Section 2: Q-Learning Simulation ==========

class QLearningAgent:
    def __init__(self, valuation_distribution=lambda: np.random.uniform(0, 1), action_min=0, action_max=1, granularity=51, epsilon=0.99, alpha=0.1, min_val=0, max_val=1, epsilon_decay=0.99999):
        self.valuation_distribution = valuation_distribution
        self.actions = action_min + (np.linspace(0, 1, granularity) ** (1/1)) * (action_max - action_min)
        self.epsilon = epsilon
        self.epsilon_initial = epsilon
        self.epsilon_min = 0.1
        self.epsilon_decay = epsilon_decay
        self.alpha = alpha
        self.min_val = min_val
        self.max_val = max_val
        self.granularity = granularity
        self.q_values = np.zeros((granularity, len(self.actions)))
        self.valuation = self.valuation_distribution()
        self.state = self.get_state(self.valuation)

    def get_state(self, valuation):
        bin_centers = np.linspace(self.min_val, self.max_val, self.granularity)
        return np.argmin(np.abs(bin_centers - valuation))
    
    def refresh_valuation(self):
        self.valuation = self.valuation_distribution()
        self.state = self.get_state(self.valuation)
    
    def choose_action(self):
        if np.random.rand() < self.epsilon:
            return np.random.choice(self.actions)
        else:
            return self.actions[np.argmax(self.q_values[self.state])]

    def update_q_values(self, chosen_action, reward):
        action_index = np.where(self.actions == chosen_action)[0][0]
        self.q_values[self.state, action_index] += self.alpha * (reward - self.q_values[self.state, action_index])

    def feedback(self, reward, action):
        self.update_q_values(action, reward)
        self.epsilon = max(self.epsilon_min, self.epsilon * self.epsilon_decay)

class SingleSidedAuction:
    def __init__(self, agents, price_floor=0, price_ceiling=1, payment_rule=None):
        self.agents = agents
        self.price_floor = price_floor
        self.price_ceiling = price_ceiling
        self.payment_rule = payment_rule

    def run_auction(self):
        for agent in self.agents: agent.refresh_valuation()
        bids = [max(self.price_floor, min(agent.choose_action(), self.price_ceiling)) for agent in self.agents]
        winner_index, bids = self.payment_rule(bids)
        
        for idx, agent in enumerate(self.agents):
            if idx == winner_index:
                payoff = agent.valuation - bids[idx]  # Winner gets valuation minus their bid
            else:
                payoff = -bids[idx]  # Losers lose their bid
            agent.feedback(payoff, bids[idx])

# ========== Section 3: All-Pay Auction Rule and Simulation ==========

def all_pay_rule(bids):
    max_bid = np.max(bids)
    winner_candidates = np.where(bids == max_bid)[0]
    winner = np.random.choice(winner_candidates) if len(winner_candidates) > 1 else winner_candidates[0]
    return winner, bids  # Return individual bids for each agent

def run_monte_carlo_simulation(agents, auction, num_simulations=2, num_rounds=10000):
    all_bids_for_valuations = [[] for _ in range(len(agents))]
    for sim in range(num_simulations):
        run_single_simulation(agents, auction, num_rounds=num_rounds)
        for i, agent in enumerate(agents):
            avg_bids_for_valuation = [agent.actions[np.argmax(agent.q_values[state])] for state in range(agent.granularity)]
            all_bids_for_valuations[i].append(avg_bids_for_valuation)

    percentiles = {}
    for i in range(len(agents)):
        avg_bids_for_agent = np.array(all_bids_for_valuations[i])
        percentiles[i] = {
            "5th": np.percentile(avg_bids_for_agent, 5, axis=0),
            "median": np.median(avg_bids_for_agent, axis=0),
            "95th": np.percentile(avg_bids_for_agent, 95, axis=0),
        }
    return percentiles

# ========== Section 4: Plotting and Comparing to All-Pay Auction ==========

def plot_all_pay_simulation(filepath, percentiles, v_guess, axes, valuation_grid, predicted_bids):
    true_valuations, true_bids = get_true_bids(filepath)

    valuations_scaled = np.linspace(0, 30, len(percentiles[0]["median"]))
    simulated_bids_scaled = 30 * percentiles[0]["median"]

    axes[1, 0].plot(valuation_grid, predicted_bids, label='Regression Line', color='red')
    axes[1, 0].scatter(true_valuations, true_bids, label='Actual Bids', color='green', s=10, alpha=0.6)
    axes[1, 0].plot(valuations_scaled, simulated_bids_scaled, label='Simulated Median Bidding Function (Scaled)', color='blue', linestyle='--')
    axes[1, 0].set_xlabel('Valuation')
    axes[1, 0].set_ylabel('Bid')
    axes[1, 0].set_title('Observed vs Simulated Bidding Functions with Actual Bids (All-Pay)')
    axes[1, 0].legend()
    axes[1, 0].set_ylim(0, 30)

def plot_kde_comparison_all_pay(filepath, percentiles, axes):
    true_valuations, true_bids = get_true_bids(filepath)

    simulated_bids_scaled = 30 * percentiles[0]["median"]

    sns.kdeplot(true_bids, label='Observed Bids', color='red', fill=True, ax=axes[1, 1])
    sns.kdeplot(simulated_bids_scaled, label='Simulated Bids', color='blue', linestyle='--', fill=True, ax=axes[1, 1])
    axes[1, 1].set_xlabel('Bid')
    axes[1, 1].set_ylabel('Density')
    axes[1, 1].set_title('KDE of Observed vs Simulated Bids (All-Pay)')
    axes[1, 1].legend()
    axes[1, 1].set_xlim(0, 30)

# ========== Example Usage for All-Pay Auction ==========

filepath = '/home/pp712/double-auctions/code/14_simple/clean_kaplan_data.csv'
agents = [QLearningAgent(epsilon=0.99, epsilon_decay=0.99999) for _ in range(2)]
auction = SingleSidedAuction(agents, payment_rule=all_pay_rule)
percentiles = run_monte_carlo_simulation(agents, auction, num_simulations=1, num_rounds=1000000)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
model, valuation_grid, predicted_bids = plot_all_pay_bids(filepath, axes)
plot_all_pay_simulation(filepath, percentiles, v_guess=30, axes=axes, valuation_grid=valuation_grid, predicted_bids=predicted_bids)
plot_kde_comparison_all_pay(filepath, percentiles, axes)
plt.tight_layout()
plt.show()
```
