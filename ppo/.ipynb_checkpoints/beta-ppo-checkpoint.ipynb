### First price

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.distributions import Beta
import gym
from gym import spaces
import matplotlib.pyplot as plt
import seaborn as sns

# Define the Auction Environment
class AuctionEnv(gym.Env):
    def __init__(self, num_other_agents=1, valuation_distribution=lambda: np.random.uniform(0, 1)):
        super(AuctionEnv, self).__init__()
        self.observation_space = spaces.Box(low=0.0, high=1.0, shape=(1,), dtype=np.float32)
        self.action_space = spaces.Box(low=0.0, high=1.0, shape=(1,), dtype=np.float32)
        self.num_other_agents = num_other_agents
        self.valuation_distribution = valuation_distribution
        self.reset()

    def reset(self):
        self.valuation = self.valuation_distribution()
        self.state = np.array([self.valuation], dtype=np.float32)
        return self.state

    def step(self, action):
        agent_bid = np.clip(action[0], 0.0, 1.0)
        other_bids = self.other_agents_bids()
        all_bids = np.append(other_bids, agent_bid)
        winner = np.argmax(all_bids)
        payment = all_bids[winner]
        if winner == len(all_bids) - 1:
            reward = self.valuation - payment
        else:
            reward = 0.0
        done = True  # Single-step episode
        info = {}
        self.reset()  # Start new auction
        return self.state, reward, done, info

    def other_agents_bids(self):
        other_valuations = [self.valuation_distribution() for _ in range(self.num_other_agents)]
        other_bids = [0.5 * v for v in other_valuations]  # Assuming truthful bidding by others
        return np.array(other_bids)

    def render(self, mode='human'):
        pass

import torch.nn.functional as F  # Import torch.nn.functional for softplus

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.distributions import Beta

class Agent(nn.Module):
    def __init__(self, observation_space, action_space):
        super(Agent, self).__init__()
        # Modify network to process correct input shape
        self.actor = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),  # Input layer matching observation size
            nn.Tanh(),
            nn.Linear(64, 64),  # Hidden layer
            nn.Tanh()
        )
        # Heads for alpha and beta
        self.alpha_head = nn.Linear(64, action_space.shape[0])
        self.beta_head = nn.Linear(64, action_space.shape[0])
        # Critic network
        self.critic = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh(),
            nn.Linear(64, 1)
        )

    def get_action(self, x):
        #print(f"Input to actor: {x.shape}")  # Debugging shape of input to the actor network
        x = self.actor(x)
        #print(f"Output from actor (x): {x.shape}")  # Debugging output of actor network

        alpha = F.softplus(self.alpha_head(x)) + 1.0  # Ensure alpha > 1
        beta = F.softplus(self.beta_head(x)) + 1.0    # Ensure beta > 1
        #print(f"Alpha shape: {alpha.shape}, Beta shape: {beta.shape}")  # Debugging shapes of alpha and beta

        dist = Beta(alpha, beta)
        action = dist.sample()
        log_prob = dist.log_prob(action).sum(dim=-1)
        #print(f"Action shape: {action.shape}, Log prob shape: {log_prob.shape}")  # Debugging action and log prob shapes
        return action, log_prob

    def get_value(self, x):
        #print(f"Input to critic: {x.shape}")  # Debugging shape of input to the critic network
        value = self.critic(x)
        #print(f"Value output shape: {value.shape}")  # Debugging output shape of critic
        return value


# Training Parameters
num_iterations = 1000  # Number of training iterations
batch_size = 1000     # Number of auctions per iteration
learning_rate = 3e-3
gamma = 0.0           # Immediate reward only
epsilon = 0.3        # PPO clipping epsilon
num_epochs = 4        # PPO epochs
num_minibatches = 4   # Number of minibatches for PPO update

# Create Environment and Agent
env = AuctionEnv()
obs_space = env.observation_space
act_space = env.action_space
agent = Agent(obs_space, act_space)
optimizer = optim.Adam(agent.parameters(), lr=learning_rate)

# Training Loop
for iteration in range(num_iterations):
    obs_batch = []
    actions_batch = []
    rewards_batch = []
    log_probs_batch = []
    values_batch = []
    
    # To track observed bids and rewards per valuation
    observed_valuations = []
    observed_bids = []
    observed_rewards = []

    obs = env.reset()
    for _ in range(batch_size):
        obs_tensor = torch.tensor(obs, dtype=torch.float32).unsqueeze(0)
        action, log_prob = agent.get_action(obs_tensor)
        value = agent.get_value(obs_tensor)

        action_np = action.detach().numpy()[0]
        obs_new, reward, done, _ = env.step(action_np)

        obs_batch.append(obs)
        actions_batch.append(action_np)
        rewards_batch.append(reward)
        log_probs_batch.append(log_prob.detach().numpy())
        values_batch.append(value.detach().numpy()[0])

        # Store observed bids and rewards
        observed_valuations.append(obs[0])  # Current valuation
        observed_bids.append(action_np)     # Corresponding bid
        observed_rewards.append(reward)     # Corresponding reward

        obs = obs_new

    # Convert batches to tensors
    obs_batch = torch.tensor(obs_batch, dtype=torch.float32)
    actions_batch = torch.tensor(actions_batch, dtype=torch.float32)
    rewards_batch = torch.tensor(rewards_batch, dtype=torch.float32)
    log_probs_batch = torch.tensor(log_probs_batch, dtype=torch.float32).squeeze()
    values_batch = torch.tensor(values_batch, dtype=torch.float32).squeeze()

    # Compute Advantages and Returns
    returns = rewards_batch  # Since gamma=0, returns are immediate rewards
    advantages = returns - values_batch.detach()

    # Normalize Advantages
    advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

    # PPO Update
    for _ in range(num_epochs):
        indices = np.arange(batch_size)
        np.random.shuffle(indices)
        minibatch_size = batch_size // num_minibatches
        for start in range(0, batch_size, minibatch_size):
            end = start + minibatch_size
            mb_inds = indices[start:end]

            mb_obs = obs_batch[mb_inds]
            mb_actions = actions_batch[mb_inds]
            mb_log_probs_old = log_probs_batch[mb_inds]
            mb_advantages = advantages[mb_inds]
            mb_returns = returns[mb_inds]

            # Forward Pass
            alpha_beta = agent.actor(mb_obs)
            alpha = F.softplus(agent.alpha_head(alpha_beta)) + 1.0
            beta = F.softplus(agent.beta_head(alpha_beta)) + 1.0
            dist = Beta(alpha, beta)
            mb_log_probs = dist.log_prob(mb_actions).sum(dim=-1)
            entropy = dist.entropy().sum(dim=-1).mean()
            values = agent.critic(mb_obs).squeeze()

            # Calculate Ratios
            ratios = (mb_log_probs - mb_log_probs_old).exp()

            # Policy Loss
            surr1 = ratios * mb_advantages
            surr2 = torch.clamp(ratios, 1 - epsilon, 1 + epsilon) * mb_advantages
            policy_loss = -torch.min(surr1, surr2).mean()

            # Value Loss
            value_loss = nn.MSELoss()(values, mb_returns)

            # Total Loss
            loss = policy_loss + 0.5 * value_loss - 0.01 * entropy

            # Backpropagation
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    # Optional: Print progress and plot every 10 iterations
    if (iteration + 1) % 10 == 0:
        avg_reward = rewards_batch.mean().item()
        print(f"Iteration {iteration + 1}/{num_iterations}, Average Reward: {avg_reward:.4f}")

        # Plotting the current learned bidding function
        valuations = np.linspace(0, 1, 100)
        bids = []
        for v in valuations:
            obs_tensor = torch.tensor([v], dtype=torch.float32).unsqueeze(0)
            with torch.no_grad():
                alpha_beta = agent.actor(obs_tensor)
                alpha = F.softplus(agent.alpha_head(alpha_beta)) + 1.0
                beta = F.softplus(agent.beta_head(alpha_beta)) + 1.0
                action_mean = alpha / (alpha + beta)  # Mean of Beta distribution
            bids.append(action_mean.numpy()[0])

        # Set up the 1x3 grid
        fig, axes = plt.subplots(1, 3, figsize=(18, 5))

        # Plot 1: Learned Bidding Function
        axes[0].plot(valuations, bids, label='Learned Bidding Function')
        axes[0].plot(valuations, 0.5 * valuations, label='Truthful Bidding', linestyle='--')
        axes[0].set_xlabel('Valuation')
        axes[0].set_ylabel('Bid')
        axes[0].set_title(f'Learned Bidding Strategy (Iteration {iteration + 1})')
        axes[0].legend()

        # 2. Plot regression of bids vs observed valuations
        sns.regplot(x=observed_valuations, y=observed_bids, scatter_kws={'s':5}, line_kws={"color":"r"}, ax=axes[1])
        axes[1].set_xlabel('Valuation (Observed)')
        axes[1].set_ylabel('Bids (Observed)')
        axes[1].set_title(f'Regression of Bids vs Valuation\nScatter: Observed Bids')

        # 3. Plot regression of rewards vs observed valuations
        sns.regplot(x=observed_valuations, y=observed_rewards, scatter_kws={'s':5}, line_kws={"color":"g"}, ax=axes[2])
        axes[2].set_xlabel('Valuation (Observed)')
        axes[2].set_ylabel('Rewards (Observed)')
        axes[2].set_title(f'Regression of Rewards vs Valuation\nScatter: Observed Rewards')
        
        # Show the plot
        plt.tight_layout()
        plt.show()
```

### all-pay

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.distributions import Beta
import gym
from gym import spaces
import matplotlib.pyplot as plt
import seaborn as sns

# Define the Auction Environment for All-Pay Auction
class AllPayAuctionEnv(gym.Env):
    def __init__(self, num_agents=2, valuation_distribution=lambda: np.random.uniform(0, 1)):
        super(AllPayAuctionEnv, self).__init__()
        self.num_agents = num_agents
        self.observation_space = spaces.Box(low=0.0, high=1.0, shape=(1,), dtype=np.float32)
        self.action_space = spaces.Box(low=0.0, high=1.0, shape=(1,), dtype=np.float32)
        self.valuation_distribution = valuation_distribution
        self.reset()

    def reset(self):
        # Each agent gets a private valuation
        self.valuations = [self.valuation_distribution() for _ in range(self.num_agents)]
        # Observations are the valuations for each agent
        self.states = [np.array([v], dtype=np.float32) for v in self.valuations]
        return self.states

    def step(self, actions):
        # actions: list of actions (bids) from each agent
        # Clip actions to be between 0 and 1 (bid must be between 0 and 1)
        bids = [np.clip(a[0], 0.0, 1.0) for a in actions]
        # Determine the winner: highest bid wins
        bids = np.array(bids)
        max_bid = np.max(bids)
        winner_indices = np.where(bids == max_bid)[0]
        # If tie, select winner randomly among those who bid the max bid
        winner = np.random.choice(winner_indices)

        # Compute rewards
        rewards = [-bids[i] for i in range(self.num_agents)]  # Losers lose their bid
        rewards[winner] = self.valuations[winner] - bids[winner]  # Winner gets valuation minus their bid

        done = True  # Single-step episode
        info = {'bids': bids, 'valuations': self.valuations, 'winner': winner}
        next_states = self.reset()
        return next_states, rewards, done, info

    def render(self, mode='human'):
        pass

# Define the Beta-PPO Agent
class PPOAgent:
    def __init__(self, observation_space, action_space, lr=3e-4):
        self.observation_space = observation_space
        self.action_space = action_space
        self.lr = lr

        # Actor network outputs alpha and beta for Beta distribution
        self.actor = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh()
        )
        self.alpha_head = nn.Linear(64, action_space.shape[0])
        self.beta_head = nn.Linear(64, action_space.shape[0])

        # Critic network
        self.critic = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh(),
            nn.Linear(64, 1)
        )

        # Optimizer for the agent
        self.optimizer = optim.Adam(self.parameters(), lr=self.lr)

    def parameters(self):
        return list(self.actor.parameters()) + list(self.alpha_head.parameters()) + list(self.beta_head.parameters()) + list(self.critic.parameters())

    def get_action(self, x):
        x = self.actor(x)
        alpha = F.softplus(self.alpha_head(x)) + 1.0  # Ensure alpha > 1
        beta = F.softplus(self.beta_head(x)) + 1.0    # Ensure beta > 1
        dist = Beta(alpha, beta)
        action = dist.sample()  # Sample action from Beta distribution
        log_prob = dist.log_prob(action).sum(dim=-1)
        return action, log_prob

    def get_value(self, x):
        return self.critic(x)

    def compute_loss(self, obs_batch, actions_batch, log_probs_old, returns, advantages, epsilon=0.2):
        # Actor loss
        x = self.actor(obs_batch)
        alpha = F.softplus(self.alpha_head(x)) + 1.0
        beta = F.softplus(self.beta_head(x)) + 1.0
        dist = Beta(alpha, beta)
        log_probs = dist.log_prob(actions_batch).sum(dim=-1)
        ratios = torch.exp(log_probs - log_probs_old)
        surr1 = ratios * advantages
        surr2 = torch.clamp(ratios, 1 - epsilon, 1 + epsilon) * advantages
        actor_loss = -torch.min(surr1, surr2).mean()

        # Critic loss
        values = self.critic(obs_batch).squeeze()
        critic_loss = nn.MSELoss()(values, returns)

        # Entropy bonus
        entropy = dist.entropy().sum(dim=-1).mean()

        # Total loss
        loss = actor_loss + 0.5 * critic_loss - 0.005 * entropy
        return loss

    def update(self, loss):
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

# Training Parameters
num_iterations = 1000  # Number of training iterations
batch_size = 1000       # Number of auctions per iteration
learning_rate = 3e-3
gamma = 0.0           # Immediate reward only
epsilon = 0.1         # PPO clipping epsilon
num_epochs = 4        # PPO epochs
num_minibatches = 4   # Number of minibatches for PPO update

# Create Environment and Agents
num_agents = 2  # All-pay auction with 2 bidders
env = AllPayAuctionEnv(num_agents=num_agents)  # Use All-Pay Auction environment
agents = [PPOAgent(env.observation_space, env.action_space, lr=learning_rate)
          for _ in range(num_agents)]

# Training Loop
for iteration in range(num_iterations):
    # Initialize data structures for each agent
    obs_batch = [[] for _ in range(num_agents)]
    actions_batch = [[] for _ in range(num_agents)]
    rewards_batch = [[] for _ in range(num_agents)]
    log_probs_batch = [[] for _ in range(num_agents)]
    values_batch = [[] for _ in range(num_agents)]
    
    # To track observed bids and rewards for all agents
    observed_valuations = [[] for _ in range(num_agents)]
    observed_bids = [[] for _ in range(num_agents)]
    observed_rewards = [[] for _ in range(num_agents)]

    # Reset environment
    states = env.reset()

    for _ in range(batch_size):
        actions = []
        for i in range(num_agents):
            obs = states[i]
            obs_tensor = torch.tensor(obs, dtype=torch.float32).unsqueeze(0)
            action, log_prob = agents[i].get_action(obs_tensor)
            value = agents[i].get_value(obs_tensor)

            action_np = action.detach().numpy()[0]
            actions.append(action_np)

            # Store data
            obs_batch[i].append(obs)
            actions_batch[i].append(action_np)
            log_probs_batch[i].append(log_prob.detach().numpy())
            values_batch[i].append(value.detach().numpy()[0])

        # Step environment with all actions
        next_states, rewards, done, _ = env.step(actions)

        # Track observed data for all agents
        for i in range(num_agents):
            observed_valuations[i].append(states[i][0])  # Valuation for agent i
            observed_bids[i].append(actions[i])          # Bid for agent i
            observed_rewards[i].append(rewards[i])       # Reward for agent i

        # Store rewards
        for i in range(num_agents):
            rewards_batch[i].append(rewards[i])

        # Update states
        states = next_states

    # PPO Update for each agent
    for i in range(num_agents):
        obs_batch_i = torch.tensor(obs_batch[i], dtype=torch.float32)
        actions_batch_i = torch.tensor(actions_batch[i], dtype=torch.float32)
        rewards_batch_i = torch.tensor(rewards_batch[i], dtype=torch.float32)
        log_probs_batch_i = torch.tensor(log_probs_batch[i], dtype=torch.float32).squeeze()
        values_batch_i = torch.tensor(values_batch[i], dtype=torch.float32).squeeze()

        # Compute Advantages and Returns
        returns = rewards_batch_i  # Since gamma=0, returns are immediate rewards
        advantages = returns - values_batch_i.detach()

        # Normalize Advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        # PPO Update
        for _ in range(num_epochs):
            indices = np.arange(batch_size)
            np.random.shuffle(indices)
            minibatch_size = batch_size // num_minibatches
            for start in range(0, batch_size, minibatch_size):
                end = start + minibatch_size
                mb_inds = indices[start:end]

                mb_obs = obs_batch_i[mb_inds]
                mb_actions = actions_batch_i[mb_inds]
                mb_log_probs_old = log_probs_batch_i[mb_inds]
                mb_advantages = advantages[mb_inds]
                mb_returns = returns[mb_inds]

                # Compute loss
                loss = agents[i].compute_loss(mb_obs, mb_actions, mb_log_probs_old, mb_returns, mb_advantages, epsilon)

                # Update agent
                agents[i].update(loss)

    # Optional: Print progress and plot every 10 iterations
    if (iteration + 1) % 10 == 0:
        avg_rewards = [np.mean(rewards_batch[i]) for i in range(num_agents)]
        print(f"Iteration {iteration + 1}/{num_iterations}, Average Rewards: {avg_rewards}")

        # Create 1x3 grid plot for both agents
        fig, axes = plt.subplots(2, 3, figsize=(18, 10))

        valuations = np.linspace(0, 1, 100)
        
        for i in range(num_agents):
            bids_agent = []
            for v in valuations:
                obs_tensor = torch.tensor([v], dtype=torch.float32).unsqueeze(0)
                with torch.no_grad():
                    x = agents[i].actor(obs_tensor)
                    alpha = F.softplus(agents[i].alpha_head(x)) + 1.0
                    beta = F.softplus(agents[i].beta_head(x)) + 1.0
                    action_mean = alpha / (alpha + beta)  # Mean of Beta distribution
                bids_agent.append(action_mean.numpy()[0])

            # 1. Plot the Learned Bidding Function for Agent i
            axes[i, 0].plot(valuations, bids_agent, label=f'Learned Bidding Function (Agent {i+1})')
            optimal_bids = valuations ** 2 / 2  # For all-pay auction, this may differ based on strategy
            axes[i, 0].plot(valuations, optimal_bids, label='Optimal Bid (b = 0.5v)', linestyle='--')
            axes[i, 0].set_xlabel('Valuation')
            axes[i, 0].set_ylabel('Bid')
            axes[i, 0].set_title(f'Learned Bidding Strategy (Agent {i+1})')
            axes[i, 0].legend()

            # 2. Plot regression of bids vs observed valuations for Agent i
            sns.regplot(x=observed_valuations[i], y=observed_bids[i], scatter_kws={'s':5}, line_kws={"color":"r"}, ax=axes[i, 1])
            axes[i, 1].set_xlabel('Valuation (Observed)')
            axes[i, 1].set_ylabel('Bids (Observed)')
            axes[i, 1].set_title(f'Regression of Bids vs Valuation (Agent {i+1})\nScatter: Observed Bids')

            # 3. Plot regression of rewards vs observed valuations for Agent i
            sns.regplot(x=observed_valuations[i], y=observed_rewards[i], scatter_kws={'s':5}, line_kws={"color":"g"}, ax=axes[i, 2])
            axes[i, 2].set_xlabel('Valuation (Observed)')
            axes[i, 2].set_ylabel('Rewards (Observed)')
            axes[i, 2].set_title(f'Regression of Rewards vs Valuation (Agent {i+1})\nScatter: Observed Rewards')

        plt.tight_layout()
        plt.show()
```

### korean

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.distributions.normal import Normal
import gym
from gym import spaces
import matplotlib.pyplot as plt
import seaborn as sns

# Define the Korean Auction Environment
class KoreanAuctionEnv(gym.Env):
    def __init__(self, num_agents=2):
        super(KoreanAuctionEnv, self).__init__()
        self.num_agents = num_agents
        # Observation space: [v, x], where v ∈ [0,1], x ∈ {0,1,2}
        self.observation_space = spaces.Box(low=np.array([0.0, 0.0]), high=np.array([1.0, 2.0]), dtype=np.float32)
        # Action space: bid ∈ [0,1]
        self.action_space = spaces.Box(low=0.0, high=1.0, shape=(1,), dtype=np.float32)
        self.reset()

    def reset(self):
        # Round counter
        self.current_round = 0
        # Each agent draws a private valuation v ∈ [0,1]
        self.valuations = [np.random.uniform(0, 1) for _ in range(self.num_agents)]
        # Initial state x = 0 for all agents
        self.states = [np.array([self.valuations[i], 0.0], dtype=np.float32) for i in range(self.num_agents)]
        # Store bids from both rounds for each agent
        self.bids_history = [[] for _ in range(self.num_agents)]
        return self.states

    def step(self, actions):
        rewards = [0.0 for _ in range(self.num_agents)]
        info = {}
        done = False

        # Collect bids
        bids = [np.clip(a[0], 0.0, 1.0) for a in actions]
        for i in range(self.num_agents):
            self.bids_history[i].append(bids[i])

        if self.current_round == 0:
            # Determine highest bidder(s) in round 0
            max_bid = np.max(bids)
            highest_bidders = [i for i, bid in enumerate(bids) if bid == max_bid]
            # Update x for each agent
            next_states = []
            for i in range(self.num_agents):
                v = self.valuations[i]
                x = 2.0 if i in highest_bidders else 1.0
                next_states.append(np.array([v, x], dtype=np.float32))
            # No rewards in round 0
            rewards = [0.0 for _ in range(self.num_agents)]
            self.current_round += 1
            done = False  # Proceed to next round
        else:
            # Round 1: Final bids are the maximum of bids from both rounds
            final_bids = [max(self.bids_history[i]) for i in range(self.num_agents)]
            # Determine the winner
            max_final_bid = np.max(final_bids)
            winner_indices = [i for i, bid in enumerate(final_bids) if bid == max_final_bid]
            winner = np.random.choice(winner_indices) if len(winner_indices) > 1 else winner_indices[0]
            # Winner pays their own final bid and gets valuation minus payment
            rewards[winner] = self.valuations[winner] - final_bids[winner]
            # Other agents get zero reward
            # Prepare next_states (not used since the episode ends)
            next_states = [np.zeros_like(self.states[i]) for i in range(self.num_agents)]
            done = True  # Episode ends after round 1

            # Info for debugging and analysis
            info = {
                'bids_history': self.bids_history,
                'final_bids': final_bids,
                'valuations': self.valuations,
                'winner': winner
            }

        return next_states, rewards, done, info

    def render(self, mode='human'):
        pass

# Define the PPO Agent
class PPOAgent(nn.Module):
    def __init__(self, observation_space, action_space, action_bound_method='clamp', lr=3e-4):
        super(PPOAgent, self).__init__()
        self.observation_space = observation_space
        self.action_space = action_space
        self.action_bound_method = action_bound_method
        self.lr = lr

        self.actor = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.ReLU(),
            nn.Linear(64, 64),
            nn.ReLU(),
            nn.Linear(64, action_space.shape[0]),
            nn.Sigmoid(),  # Output between 0 and 1
        )
        self.critic = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.ReLU(),
            nn.Linear(64, 64),
            nn.ReLU(),
            nn.Linear(64, 1)
        )
        self.log_std = nn.Parameter(torch.zeros(action_space.shape[0]))

        self.optimizer = optim.Adam(self.parameters(), lr=self.lr)

    def get_action(self, x):
        mean = self.actor(x)  # Output between 0 and 1
        std = self.log_std.exp()
        dist = Normal(mean, std)
        action = dist.sample()
        if self.action_bound_method == 'clamp':
            action = torch.clamp(action, 0.0, 1.0)
        log_prob = dist.log_prob(action).sum(dim=-1)
        return action, log_prob

    def get_value(self, x):
        return self.critic(x)

    def compute_loss(self, obs_batch, actions_batch, log_probs_old, returns, advantages, epsilon=0.2):
        # Actor loss
        mean = self.actor(obs_batch)
        std = self.log_std.exp()
        dist = Normal(mean, std)
        actions_bound = torch.clamp(actions_batch, 0.0, 1.0)
        log_probs = dist.log_prob(actions_bound).sum(dim=-1)
        ratios = torch.exp(log_probs - log_probs_old)
        surr1 = ratios * advantages
        surr2 = torch.clamp(ratios, 1 - epsilon, 1 + epsilon) * advantages
        actor_loss = -torch.min(surr1, surr2).mean()
        # Critic loss
        values = self.critic(obs_batch).squeeze()
        critic_loss = nn.MSELoss()(values, returns)
        # Entropy bonus
        entropy = dist.entropy().sum(dim=-1).mean()
        # Total loss
        loss = actor_loss + 0.5 * critic_loss - 0.005 * entropy
        return loss

    def update(self, loss):
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

# Training Parameters
num_iterations = 1000  # Number of training iterations
batch_size = 1000       # Number of episodes per iteration
learning_rate = 3e-3
gamma = 1.0           # Full discounting
epsilon = 0.1         # PPO clipping epsilon
num_epochs = 4        # PPO epochs
num_minibatches = 4   # Number of minibatches for PPO update

# Create Environment and Agents for the Korean Auction
env = KoreanAuctionEnv(num_agents=2)
agents = [PPOAgent(env.observation_space, env.action_space, action_bound_method='clamp', lr=learning_rate)
          for _ in range(env.num_agents)]

# Training Loop
for iteration in range(num_iterations):
    # Initialize data structures for each agent
    obs_batch = [[] for _ in range(env.num_agents)]
    actions_batch = [[] for _ in range(env.num_agents)]
    rewards_batch = [[] for _ in range(env.num_agents)]
    log_probs_batch = [[] for _ in range(env.num_agents)]
    values_batch = [[] for _ in range(env.num_agents)]
    dones_batch = [[] for _ in range(env.num_agents)]

    # For plotting
    observed_states = [[] for _ in range(env.num_agents)]
    observed_bids = [[] for _ in range(env.num_agents)]
    observed_rewards = [[] for _ in range(env.num_agents)]

    for episode in range(batch_size):
        # Reset environment
        states = env.reset()
        done = False
        step = 0

        while not done:
            actions = []
            for i in range(env.num_agents):
                obs = states[i]
                obs_tensor = torch.tensor(obs, dtype=torch.float32).unsqueeze(0)
                action, log_prob = agents[i].get_action(obs_tensor)
                value = agents[i].get_value(obs_tensor)

                action_np = action.detach().numpy()[0]
                actions.append(action_np)

                # Store data
                obs_batch[i].append(obs)
                actions_batch[i].append(action_np)
                log_probs_batch[i].append(log_prob.detach().numpy())
                values_batch[i].append(value.detach().numpy()[0])
                dones_batch[i].append(done)

                # For plotting
                observed_states[i].append(obs)
                if step == env.current_round:
                    observed_bids[i].append(action_np)

            # Step environment with all actions
            next_states, rewards, done, info = env.step(actions)

            # Store rewards
            for i in range(env.num_agents):
                rewards_batch[i].append(rewards[i])
                observed_rewards[i].append(rewards[i])

            # Update states
            states = next_states
            step += 1

    # PPO Update for each agent
    for i in range(env.num_agents):
        obs_batch_i = torch.tensor(obs_batch[i], dtype=torch.float32)
        actions_batch_i = torch.tensor(actions_batch[i], dtype=torch.float32)
        rewards_batch_i = torch.tensor(rewards_batch[i], dtype=torch.float32)
        log_probs_batch_i = torch.tensor(log_probs_batch[i], dtype=torch.float32).squeeze()
        values_batch_i = torch.tensor(values_batch[i], dtype=torch.float32).squeeze()
        dones_batch_i = torch.tensor(dones_batch[i], dtype=torch.float32)

        # Compute Returns and Advantages
        returns = []
        G = 0
        for reward, done_flag in zip(reversed(rewards_batch_i), reversed(dones_batch_i)):
            if done_flag:
                G = 0
            G = reward + gamma * G
            returns.insert(0, G)
        returns = torch.tensor(returns, dtype=torch.float32)
        advantages = returns - values_batch_i.detach()

        # Normalize Advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        # PPO Update
        for _ in range(num_epochs):
            indices = np.arange(len(obs_batch_i))
            np.random.shuffle(indices)
            minibatch_size = len(obs_batch_i) // num_minibatches
            for start in range(0, len(obs_batch_i), minibatch_size):
                end = start + minibatch_size
                mb_inds = indices[start:end]

                mb_obs = obs_batch_i[mb_inds]
                mb_actions = actions_batch_i[mb_inds]
                mb_log_probs_old = log_probs_batch_i[mb_inds]
                mb_advantages = advantages[mb_inds]
                mb_returns = returns[mb_inds]

                # Compute loss
                loss = agents[i].compute_loss(mb_obs, mb_actions, mb_log_probs_old, mb_returns, mb_advantages, epsilon)

                # Update agent
                agents[i].update(loss)

    # Optional: Print progress and plot every 10 iterations
    if (iteration + 1) % 10 == 0:
        avg_rewards = [np.mean(rewards_batch[i]) for i in range(env.num_agents)]
        print(f"Iteration {iteration + 1}/{num_iterations}, Average Rewards: {avg_rewards}")

        # Create 2x3 grid plot for all agents
        fig, axes = plt.subplots(env.num_agents, 3, figsize=(18, 5 * env.num_agents))

        valuations = np.linspace(0, 1, 100)
        x_values = [0.0, 1.0, 2.0]  # Possible x values

        for i in range(env.num_agents):
            # 1. Plot the Learned Bidding Function for Agent i at different x
            for x in x_values:
                bids_agent = []
                for v in valuations:
                    obs_tensor = torch.tensor([v, x], dtype=torch.float32).unsqueeze(0)
                    with torch.no_grad():
                        mean = agents[i].actor(obs_tensor)
                        bid = torch.clamp(mean, 0.0, 1.0).numpy()[0][0]
                    bids_agent.append(bid)
                axes[i, 0].plot(valuations, bids_agent, label=f'x={x}')
            axes[i, 0].set_xlabel('Valuation')
            axes[i, 0].set_ylabel('Bid')
            axes[i, 0].set_title(f'Learned Bidding Strategy (Agent {i+1})')
            axes[i, 0].legend()

            # 2. Plot regression of bids vs observed valuations for Agent i
            observed_vs = [state[0] for state in observed_states[i]]
            sns.regplot(x=observed_vs, y=observed_bids[i], scatter_kws={'s':5}, line_kws={"color":"r"}, ax=axes[i, 1])
            axes[i, 1].set_xlabel('Valuation (Observed)')
            axes[i, 1].set_ylabel('Bids (Observed)')
            axes[i, 1].set_xlim(0, 1)
            axes[i, 1].set_ylim(0, 1)
            axes[i, 1].set_title(f'Regression of Bids vs Valuation (Agent {i+1})\nScatter: Observed Bids')

            # 3. Plot regression of rewards vs observed valuations for Agent i
            sns.regplot(x=observed_vs, y=observed_rewards[i], scatter_kws={'s':5}, line_kws={"color":"g"}, ax=axes[i, 2])
            axes[i, 2].set_xlabel('Valuation (Observed)')
            axes[i, 2].set_ylabel('Rewards (Observed)')
            axes[i, 2].set_xlim(0, 1)
            axes[i, 2].set_title(f'Regression of Rewards vs Valuation (Agent {i+1})\nScatter: Observed Rewards')

        plt.tight_layout()
        plt.show()
```

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.distributions import Beta
import gym
from gym import spaces
import matplotlib.pyplot as plt
import seaborn as sns
import torch.nn.functional as F  # For softplus activation

# Define the Korean Auction Environment
class KoreanAuctionEnv(gym.Env):
    def __init__(self, num_agents=2):
        super(KoreanAuctionEnv, self).__init__()
        self.num_agents = num_agents
        # Observation space: [v, x], where v ∈ [0,1], x ∈ {0,1,2}
        self.observation_space = spaces.Box(low=np.array([0.0, 0.0]), high=np.array([1.0, 2.0]), dtype=np.float32)
        # Action space: bid ∈ [0,1]
        self.action_space = spaces.Box(low=0.0, high=1.0, shape=(1,), dtype=np.float32)
        self.reset()

    def reset(self):
        # Round counter
        self.current_round = 0
        # Each agent draws a private valuation v ∈ [0,1]
        self.valuations = [np.random.uniform(0, 1) for _ in range(self.num_agents)]
        # Initial state x = 0 for all agents
        self.states = [np.array([self.valuations[i], 0.0], dtype=np.float32) for i in range(self.num_agents)]
        # Store bids from both rounds for each agent
        self.bids_history = [[] for _ in range(self.num_agents)]
        return self.states

    def step(self, actions):
        rewards = [0.0 for _ in range(self.num_agents)]
        info = {}
        done = False

        # Collect bids
        bids = [np.clip(a[0], 0.0, 1.0) for a in actions]
        for i in range(self.num_agents):
            self.bids_history[i].append(bids[i])

        if self.current_round == 0:
            # Determine highest bidder(s) in round 0
            max_bid = np.max(bids)
            highest_bidders = [i for i, bid in enumerate(bids) if bid == max_bid]
            # Update x for each agent
            next_states = []
            for i in range(self.num_agents):
                v = self.valuations[i]
                x = 2.0 if i in highest_bidders else 1.0
                next_states.append(np.array([v, x], dtype=np.float32))
            # No rewards in round 0
            rewards = [0.0 for _ in range(self.num_agents)]
            self.current_round += 1
            done = False  # Proceed to next round
        else:
            # Round 1: Final bids are the maximum of bids from both rounds
            final_bids = [max(self.bids_history[i]) for i in range(self.num_agents)]
            # Determine the winner
            max_final_bid = np.max(final_bids)
            winner_indices = [i for i, bid in enumerate(final_bids) if bid == max_final_bid]
            winner = np.random.choice(winner_indices) if len(winner_indices) > 1 else winner_indices[0]
            # Winner pays their own final bid and gets valuation minus payment
            rewards[winner] = self.valuations[winner] - final_bids[winner]
            # Other agents get zero reward
            # Prepare next_states (not used since the episode ends)
            next_states = [np.zeros_like(self.states[i]) for i in range(self.num_agents)]
            done = True  # Episode ends after round 1

            # Info for debugging and analysis
            info = {
                'bids_history': self.bids_history,
                'final_bids': final_bids,
                'valuations': self.valuations,
                'winner': winner
            }

        return next_states, rewards, done, info

    def render(self, mode='human'):
        pass

# Define the PPO Agent with Beta Distribution
class PPOAgent(nn.Module):
    def __init__(self, observation_space, action_space, lr=3e-4):
        super(PPOAgent, self).__init__()
        self.observation_space = observation_space
        self.action_space = action_space
        self.lr = lr

        # Actor Network
        self.actor_base = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh()
            nn.Tanh(),
            nn.Linear(64, 64),
        )
        # Heads for alpha and beta parameters of the Beta distribution
        self.alpha_head = nn.Linear(64, action_space.shape[0])
        self.beta_head = nn.Linear(64, action_space.shape[0])

        # Critic Network
        self.critic = nn.Sequential(
            nn.Linear(observation_space.shape[0], 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh(),
            nn.Linear(64, 64),
            nn.Tanh(),
            nn.Linear(64, 1)
        )

        self.optimizer = optim.Adam(self.parameters(), lr=self.lr)

    def get_action(self, x):
        base = self.actor_base(x)
        alpha = F.softplus(self.alpha_head(base)) + 1.0  # Ensure alpha > 1
        beta = F.softplus(self.beta_head(base)) + 1.0    # Ensure beta > 1
        dist = Beta(alpha, beta)
        action = dist.sample()
        action = torch.clamp(action, 0.0, 1.0)  # Ensure action is within bounds
        log_prob = dist.log_prob(action).sum(dim=-1)
        return action, log_prob

    def get_value(self, x):
        return self.critic(x)

    def compute_loss(self, obs_batch, actions_batch, log_probs_old, returns, advantages, epsilon=0.2):
        # Actor loss
        base = self.actor_base(obs_batch)
        alpha = F.softplus(self.alpha_head(base)) + 1.0
        beta = F.softplus(self.beta_head(base)) + 1.0
        dist = Beta(alpha, beta)
        actions_clamped = torch.clamp(actions_batch, 0.0 + 1e-6, 1.0 - 1e-6)  # Avoid boundary issues
        log_probs = dist.log_prob(actions_clamped).sum(dim=-1)
        ratios = torch.exp(log_probs - log_probs_old)
        surr1 = ratios * advantages
        surr2 = torch.clamp(ratios, 1 - epsilon, 1 + epsilon) * advantages
        actor_loss = -torch.min(surr1, surr2).mean()
        # Critic loss
        values = self.critic(obs_batch).squeeze()
        critic_loss = nn.MSELoss()(values, returns)
        # Entropy bonus
        entropy = dist.entropy().sum(dim=-1).mean()
        # Total loss
        loss = actor_loss + 0.5 * critic_loss - 0.005 * entropy
        return loss

    def update(self, loss):
        self.optimizer.zero_grad()
        loss.backward()
        self.optimizer.step()

# Training Parameters
num_iterations = 1000  # Number of training iterations
batch_size = 1000       # Number of episodes per iteration
learning_rate = 3e-3
gamma = 1.0           # Full discounting
epsilon = 0.1         # PPO clipping epsilon
num_epochs = 4        # PPO epochs
num_minibatches = 4   # Number of minibatches for PPO update

# Create Environment and Agents for the Korean Auction
env = KoreanAuctionEnv(num_agents=2)
agents = [PPOAgent(env.observation_space, env.action_space, lr=learning_rate) for _ in range(env.num_agents)]

# Training Loop
for iteration in range(num_iterations):
    # Initialize data structures for each agent
    obs_batch = [[] for _ in range(env.num_agents)]
    actions_batch = [[] for _ in range(env.num_agents)]
    rewards_batch = [[] for _ in range(env.num_agents)]
    log_probs_batch = [[] for _ in range(env.num_agents)]
    values_batch = [[] for _ in range(env.num_agents)]
    dones_batch = [[] for _ in range(env.num_agents)]

    # For plotting
    observed_states = [[] for _ in range(env.num_agents)]
    observed_bids = [[] for _ in range(env.num_agents)]
    observed_rewards = [[] for _ in range(env.num_agents)]

    for episode in range(batch_size):
        # Reset environment
        states = env.reset()
        done = False
        step = 0

        while not done:
            actions = []
            for i in range(env.num_agents):
                obs = states[i]
                obs_tensor = torch.tensor(obs, dtype=torch.float32).unsqueeze(0)
                action, log_prob = agents[i].get_action(obs_tensor)
                value = agents[i].get_value(obs_tensor)

                action_np = action.detach().numpy()[0]
                actions.append(action_np)

                # Store data
                obs_batch[i].append(obs)
                actions_batch[i].append(action_np)
                log_probs_batch[i].append(log_prob.detach().numpy())
                values_batch[i].append(value.detach().numpy()[0])
                dones_batch[i].append(done)

                # For plotting
                observed_states[i].append(obs)
                if step == env.current_round:
                    observed_bids[i].append(action_np)

            # Step environment with all actions
            next_states, rewards, done, info = env.step(actions)

            # Store rewards
            for i in range(env.num_agents):
                rewards_batch[i].append(rewards[i])
                observed_rewards[i].append(rewards[i])

            # Update states
            states = next_states
            step += 1

    # PPO Update for each agent
    for i in range(env.num_agents):
        obs_batch_i = torch.tensor(obs_batch[i], dtype=torch.float32)
        actions_batch_i = torch.tensor(actions_batch[i], dtype=torch.float32)
        rewards_batch_i = torch.tensor(rewards_batch[i], dtype=torch.float32)
        log_probs_batch_i = torch.tensor(log_probs_batch[i], dtype=torch.float32).squeeze()
        values_batch_i = torch.tensor(values_batch[i], dtype=torch.float32).squeeze()
        dones_batch_i = torch.tensor(dones_batch[i], dtype=torch.float32)

        # Compute Returns and Advantages
        returns = []
        G = 0
        for reward, done_flag in zip(reversed(rewards_batch_i), reversed(dones_batch_i)):
            if done_flag:
                G = 0
            G = reward + gamma * G
            returns.insert(0, G)
        returns = torch.tensor(returns, dtype=torch.float32)
        advantages = returns - values_batch_i.detach()

        # Normalize Advantages
        advantages = (advantages - advantages.mean()) / (advantages.std() + 1e-8)

        # PPO Update
        for _ in range(num_epochs):
            indices = np.arange(len(obs_batch_i))
            np.random.shuffle(indices)
            minibatch_size = len(obs_batch_i) // num_minibatches
            for start in range(0, len(obs_batch_i), minibatch_size):
                end = start + minibatch_size
                mb_inds = indices[start:end]

                mb_obs = obs_batch_i[mb_inds]
                mb_actions = actions_batch_i[mb_inds]
                mb_log_probs_old = log_probs_batch_i[mb_inds]
                mb_advantages = advantages[mb_inds]
                mb_returns = returns[mb_inds]

                # Compute loss
                loss = agents[i].compute_loss(mb_obs, mb_actions, mb_log_probs_old, mb_returns, mb_advantages, epsilon)

                # Update agent
                agents[i].update(loss)

    # Optional: Print progress and plot every 10 iterations
    if (iteration + 1) % 10 == 0:
        avg_rewards = [np.mean(rewards_batch[i]) for i in range(env.num_agents)]
        print(f"Iteration {iteration + 1}/{num_iterations}, Average Rewards: {avg_rewards}")

        # Create 2x3 grid plot for all agents
        fig, axes = plt.subplots(env.num_agents, 3, figsize=(18, 5 * env.num_agents))

        valuations = np.linspace(0, 1, 100)
        x_values = [0.0, 1.0, 2.0]  # Possible x values

        for i in range(env.num_agents):
            # 1. Plot the Learned Bidding Function for Agent i at different x
            for x in x_values:
                bids_agent = []
                for v in valuations:
                    obs_tensor = torch.tensor([v, x], dtype=torch.float32).unsqueeze(0)
                    with torch.no_grad():
                        base = agents[i].actor_base(obs_tensor)
                        alpha = F.softplus(agents[i].alpha_head(base)) + 1.0
                        beta = F.softplus(agents[i].beta_head(base)) + 1.0
                        action_mean = alpha / (alpha + beta)  # Mean of Beta distribution
                    bids_agent.append(action_mean.numpy()[0])
                axes[i, 0].plot(valuations, bids_agent, label=f'x={x}')
            axes[i, 0].set_xlabel('Valuation')
            axes[i, 0].set_ylabel('Bid')
            axes[i, 0].set_title(f'Learned Bidding Strategy (Agent {i+1})')
            axes[i, 0].legend()

            # 2. Plot regression of bids vs observed valuations for Agent i
            observed_vs = [state[0] for state in observed_states[i]]
            sns.regplot(x=observed_vs, y=observed_bids[i], scatter_kws={'s':5}, line_kws={"color":"r"}, ax=axes[i, 1])
            axes[i, 1].set_xlabel('Valuation (Observed)')
            axes[i, 1].set_ylabel('Bids (Observed)')
            axes[i, 1].set_xlim(0, 1)
            axes[i, 1].set_ylim(0, 1)
            axes[i, 1].set_title(f'Regression of Bids vs Valuation (Agent {i+1})\nScatter: Observed Bids')

            # 3. Plot regression of rewards vs observed valuations for Agent i
            sns.regplot(x=observed_vs, y=observed_rewards[i], scatter_kws={'s':5}, line_kws={"color":"g"}, ax=axes[i, 2])
            axes[i, 2].set_xlabel('Valuation (Observed)')
            axes[i, 2].set_ylabel('Rewards (Observed)')
            axes[i, 2].set_xlim(0, 1)
            axes[i, 2].set_title(f'Regression of Rewards vs Valuation (Agent {i+1})\nScatter: Observed Rewards')

        plt.tight_layout()
        plt.show()
```
